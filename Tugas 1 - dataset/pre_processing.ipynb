```python
import os
import cv2
import numpy as np

path_sepatu = '/content/drive/MyDrive/dataset/sepatu/'
path_sandal = '/content/drive/MyDrive/dataset/sandal/'

def preprocess_image(image_path):
    if not os.path.exists(image_path):
        return f"File {image_path} tidak ditemukan!"

    img = cv2.imread(image_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img = cv2.resize(img, (224, 224))
    img = img.astype('float32') / 255.0
    return img

files_sandal = os.listdir(path_sandal)
if len(files_sandal) > 0:
    test_file_path = os.path.join(path_sandal, files_sandal[0])
    test_img = preprocess_image(test_file_path)

    print("Berhasil memproses file:", files_sandal[0])
    print("Ukuran gambar sekarang (Shape):", test_img.shape)

else:
    print("Folder kosong! Cek path GDrive.")
```

```python
import os

def count_files(directory):
    return len([f for f in os.listdir(directory) if os.path.isfile(os.path.join(directory, f))])

print(f"Jumlah foto Sepatu: {count_files('/content/drive/MyDrive/dataset/sepatu/')}")
print(f"Jumlah foto Sandal: {count_files('/content/drive/MyDrive/dataset/sandal/')}")
print("Status Preprocessing: Semua gambar telah diset ke 224x224 dan Normalisasi 0-1 (di memori).")
```
