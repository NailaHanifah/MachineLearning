```python
import os
import cv2
import numpy as np

# 1. Pastikan Path Benar (Ganti dengan nama folder asli di Drive-mu)
path_sepatu = '/content/drive/MyDrive/dataset/sepatu/'
path_sandal = '/content/drive/MyDrive/dataset/sandal/'

def preprocess_image(image_path):
    if not os.path.exists(image_path):
        return f"File {image_path} tidak ditemukan!"

    img = cv2.imread(image_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img = cv2.resize(img, (224, 224))
    img = img.astype('float32') / 255.0
    return img

# 2. Ambil satu file secara otomatis dari folder sandal untuk test
files_sandal = os.listdir(path_sandal)
if len(files_sandal) > 0:
    test_file_path = os.path.join(path_sandal, files_sandal[0])
    test_img = preprocess_image(test_file_path)

    print("Berhasil memproses file:", files_sandal[0])
    print("Ukuran gambar sekarang (Shape):", test_img.shape)
    # Output yang benar: (224, 224, 3)
else:
    print("Folder kosong! Cek path GDrive kamu.")
```
